```python
# ============================================
# تحويل صوت المحاضرة إلى نص (Speech to Text)
# باستخدام (Whisper large-v3)
# ============================================

# الخطوة 1: تثبيت المكتبات المطلوبة
!pip install -q openai-whisper

# الخطوة 2: رفع ملف الفيديو أو الصوت
from google.colab import files
print("ارفع ملف المحاضرة اللي نزلته (mp4 أو mp3 أو wav)")
uploaded = files.upload()
audio_filename = list(uploaded.keys())[0]

# الخطوة 3: تحميل موديل (Whisper)
import whisper
print("جاري تحميل الموديل... ممكن ياخد شوية وقت أول مرة")
model = whisper.load_model("large-v3")

# الخطوة 4: التفريغ الصوتي (Transcription) مع الطوابع الزمنية (Timestamps)
print("جاري تفريغ الصوت... استنى شوية حسب مدة الملف")
result = model.transcribe(
    audio_filename,
    language="ar",
    verbose=False,
    word_timestamps=True
)

# الخطوة 5: عرض النص الكامل
print("\n===== النص الكامل =====\n")
print(result["text"])

# الخطوة 6: عرض النص مقسم بالطوابع الزمنية (Segments)
print("\n===== النص مقسم بالوقت =====\n")
for segment in result["segments"]:
    start = round(segment["start"], 1)
    end = round(segment["end"], 1)
    text = segment["text"]
    print(f"[{start}s -> {end}s] {text}")

# الخطوة 7: حفظ النتيجة في ملف (JSON)
import json
output_filename = audio_filename.rsplit(".", 1)[0] + "_transcript.json"

transcript_data = {
    "full_text": result["text"],
    "segments": [
        {
            "start": seg["start"],
            "end": seg["end"],
            "text": seg["text"]
        }
        for seg in result["segments"]
    ]
}

with open(output_filename, "w", encoding="utf-8") as f:
    json.dump(transcript_data, f, ensure_ascii=False, indent=2)

print(f"\nتم الحفظ في: {output_filename}")

# الخطوة 8: تنزيل الملف على جهازك
files.download(output_filename)
```
